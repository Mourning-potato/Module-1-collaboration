```python
# Based on testing the number of digits up to 9 digits, every number of digits after 3 digits has a number that
# when multiplied by 4 is equal to it's inverse. 9 digits even has two numbers that fit this criteria

# All of these numbers has so far started with 21, and ended with 78

# I have used a shortcut to only check the number starting with 1 or 2 to save time

# The cpu time to run the code increases exponentially as more digits are added, sometimes taking 30 seconds,
# which is why I stopped testing after 9 digits

import time

start_cpu_time = time.process_time() # checking the starting cpu time

#number = int(input("Input a 4 digit number")) # inputting a 4 digit number, making sure to change it to an int

def valid_reverse_number_finder(digits):

    count = 1
    number = str(1)
    while count < digits: # this code adds zeros until the number has as many digits as were specified in the digits
        number += str(0)
        count += 1

    number = int(number) # coverting the number back to an int

    valid_numbers = [] # list that will contain any valid numbers

    limit = number * 3 # limit for the valid numbers is always between number * 1, and number * 3

    while(number < limit):

        reversed_number = str(number)[::-1] # reversing the number

        multiplied_number = number * 4 # multiplying the original number
        if reversed_number == str(multiplied_number): # comparing the reversed number to the multiplied number
            valid_numbers.append(number)
        #else:
        #    print(reversed_number + " =/= " + str(multiplied_number))

        number += 1

    print("Valid " + str(digits) + " digit numbers: ")

    for numbers in valid_numbers:
        print(numbers)
    if not valid_numbers:
        print("There are no valid numbers")


digits = int(input("Input the number of digits you want to count up to"))

valid_reverse_number_finder(digits)

end_cpu_time = time.process_time() # recording the ending cpu time

cpu_time_consumed = end_cpu_time - start_cpu_time # finding the difference between the starting cpu time and ending cpu time
print(f"CPU time consumed: {cpu_time_consumed:.6f} seconds") # code for cpu time was found on the internet
```
