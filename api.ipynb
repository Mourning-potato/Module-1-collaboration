```python
from flask import Flask, jsonify

app = Flask(__name__)

from flask_sqlalchemy import SQLAlchemy


app.config['SQLALCHEMY_DATABASE_URI'] = 'sqlite:///data.db'
db = SQLAlchemy(app)


class Book(db.Model):
    id = db.Column(db.Integer, primary_key=True)
    name = db.Column(db.String(80), unique=True, nullable=False)
    description = db.Column(db.String(120))

    def __repr__(self):
        return f"{self.name} - {self.description}"

@app.route('/')
def index():
    return 'Hello!'

@app.route('/book')
def get_book():
    books = Book.query.all()
    output = []
    for book in books:
        book_data = {'id': book.id_num, 'book_name': book.book_name, 'author': book.author, 'publisher': book.publisher,}
        output.append(book_data)
    return {'book': output}

@app.route('/books/<id>')
def get_books(id):
    book = Book.query.get_or_404(id)
    return jsonify({'id': book.id_num, 'book_name': book.book_name, 'author': book.author, 'publisher': book.publisher,})
```
